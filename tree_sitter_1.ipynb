```python
from tree_sitter import Parser, Language
import tree_sitter_python

PYTHON_LANGUAGE = Language(tree_sitter_python.language())
parser = Parser(language=PYTHON_LANGUAGE)

code = b"""
def add(a, b):
    return a + b

def multiply(x, y):
    return x * y
"""

tree = parser.parse(code)

print("Root node:", tree.root_node)
print("Tree-sitter is working!")
```

```python
def extract_functions(node, code):
    functions = []

    if node.type == "function_definition":
        function_code = code[node.start_byte:node.end_byte].decode("utf-8")
        functions.append(function_code)

    for child in node.children:
        functions.extend(extract_functions(child, code))

    return functions


functions = extract_functions(tree.root_node, code)

for i, function in enumerate(functions, 1):
    print(f"\nFunction {i}:")
    print(function)
```

```python
from tree_sitter import Parser, Language
import tree_sitter_python

PYTHON_LANGUAGE = Language(tree_sitter_python.language())
parser = Parser(language=PYTHON_LANGUAGE)

code = b"""
def calculate_total(price, quantity):
    total = price * quantity

    if total > 1000:
        discount = total * 0.10
        total = total - discount
    else:
        discount = 0

    return total


result = calculate_total(500, 3)
print(result)
"""

tree = parser.parse(code)

print("===== ORIGINAL CODE ===telek")
print(code.decode())

print("\n===== TREE-SITTER AST ===telek")
print(tree.root_node)
```

```python
def show_tree(node, indent="", is_last=True):
    branch = "└── " if is_last else "├── "
    print(indent + branch + node.type)

    children = node.children

    for i, child in enumerate(children):
        last = i == len(children) - 1

        new_indent = indent + ("    " if is_last else "│   ")

        show_tree(child, new_indent, last)


print("===== AST TREE =====")
show_tree(tree.root_node)
```

```python
import tree_sitter_java
from tree_sitter import Parser, Language

JAVA_LANGUAGE = Language(tree_sitter_java.language())
java_parser = Parser(language=JAVA_LANGUAGE)

java_code = b"""
public class Calculator {

    public int add(int a, int b) {
        int result = a + b;
        return result;
    }

    public int multiply(int x, int y) {
        int result = x * y;
        return result;
    }

    public static void main(String[] args) {
        Calculator calc = new Calculator();
        int answer = calc.add(10, 20);
        System.out.println(answer);
    }
}
"""

java_tree = java_parser.parse(java_code);

print("===== ORIGINAL JAVA CODE ====")
print(java_code.decode())

print("\n===== TREE-SITTER AST ====")
print(java_tree.root_node)
```

```python
def show_tree(node, indent="", is_last=True):
    branch = "└── " if is_last else "├── "
    print(indent + branch + node.type)

    children = node.children

    for i, child in enumerate(children):
        last = i == len(children) - 1

        if is_last:
            new_indent = indent + "    "
        else:
            new_indent = indent + "│   "

        show_tree(child, new_indent, last)


print("===== JAVA AST TREE =====")
show_tree(java_tree.root_node)
```

```python
def extract_methods(node, code):
    methods = []

    if node.type == "method_declaration":
        method = code[node.start_byte:node.end_byte].decode(
            "utf-8", errors="ignore"
        )
        methods.append(method)

    for child in node.children:
        methods.extend(extract_methods(child, code))

    return methods


methods = extract_methods(java_tree.root_node, java_code)

print("Total Java methods:", len(methods))

for i, method in enumerate(methods, 1):
    print(f"\n===== METHOD {i} =====")
    print(method)
```
