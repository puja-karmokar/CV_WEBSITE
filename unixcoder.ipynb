```python
import torch
from unixcoder import UniXcoder

device = torch.device("cuda" if torch.cuda.is_available() else "cpu")
model = UniXcoder("microsoft/unixcoder-base")
model.to(device)
```

```python
import torch

# Two random code snippets
code1 = "def add(a, b): return a + b"
code2 = "def multiply(a, b): return a * b"

# Tokenize
tokens1 = model.tokenize([code1])
tokens2 = model.tokenize([code2])

# Get embeddings
with torch.no_grad():
    emb1 = model(torch.tensor(tokens1))[0]      # shape: (1, seq_len, hidden)
    emb2 = model(torch.tensor(tokens2))[0]

# Simple mean pooling over the sequence
vec1 = emb1.mean(dim=1).squeeze()  # shape: (hidden,)
vec2 = emb2.mean(dim=1).squeeze()

# Cosine similarity
cos_sim = torch.nn.functional.cosine_similarity(vec1.unsqueeze(0), vec2.unsqueeze(0))
print("Cosine similarity between code1 and code2:", cos_sim.item())
```

```python
import torch
from unixcoder import UniXcoder

device = torch.device("cuda" if torch.cuda.is_available() else "cpu")
model = UniXcoder("microsoft/unixcoder-base")
model.to(device)
```

```python
# দুটি Python কোড স্নিপেট (একই কাজ, ভিন্ন implementation)
code1 = """
def add(a, b):
    return a + b
"""

code2 = """
def sum_two_numbers(x, y):
    return x + y
"""

def get_embedding(code_str):
    tokens = model.tokenize([code_str])
    with torch.no_grad():
        emb = model(torch.tensor(tokens))[0]          # (1, seq_len, hidden)
    vec = emb.mean(dim=1).squeeze()     # mean pooling
    return vec

vec1 = get_embedding(code1)
vec2 = get_embedding(code2)

cos_sim = torch.nn.functional.cosine_similarity(vec1.unsqueeze(0), vec2.unsqueeze(0))
print("Cosine similarity (clone detection):", cos_sim.item())
```

```python
# Python code
code_py = """
def add(a, b):
    return a + b
"""

# JavaScript code (same logic)
code_js = """
function add(a, b) {
    return a + b;
}
"""

def get_embedding(code_str):
    tokens = model.tokenize([code_str])
    with torch.no_grad():
        emb = model(torch.tensor(tokens))[0]
    vec = emb.mean(dim=1).squeeze()
    return vec

vec_py = get_embedding(code_py)
vec_js = get_embedding(code_js)

cos_sim_cross = torch.nn.functional.cosine_similarity(vec_py.unsqueeze(0), vec_js.unsqueeze(0))
print("Cosine similarity (cross-language):", cos_sim_cross.item())
```

```python
# Python code A
python_code_a = """
def add(a, b):
    return a + b
"""

# Python code B (same logic, different style/name)
python_code_b = """
def sum_two_numbers(x, y):
    return x + y
"""

vec_py_a = get_embedding(python_code_a)
vec_py_b = get_embedding(python_code_b)

sim_python = torch.nn.functional.cosine_similarity(
    vec_py_a.unsqueeze(0),
    vec_py_b.unsqueeze(0)
)

print("Python A ↔ Python B similarity:", sim_python.item())
```

```python
# Java code A
java_code_a = """
public int add(int a, int b) {
    return a + b;
}
"""

# Java code B (same logic, different style/name)
java_code_b = """
public int sumTwoNumbers(int x, int y) {
    return x + y;
}
"""

vec_java_a = get_embedding(java_code_a)
vec_java_b = get_embedding(java_code_b)

sim_java = torch.nn.functional.cosine_similarity(
    vec_java_a.unsqueeze(0),
    vec_java_b.unsqueeze(0)
)

print("Java A ↔ Java B similarity:", sim_java.item())
```

```python
# Use same Python and Java snippets as above
vec_py = get_embedding(python_code_a)
vec_java = get_embedding(java_code_a)

sim_cross = torch.nn.functional.cosine_similarity(
    vec_py.unsqueeze(0),
    vec_java.unsqueeze(0)
)

print("Python ↔ Java cross-language similarity:", sim_cross.item())
```

```python
# Ensure unixcoder.py is available and correctly named
!rm -f unixcoder.py unixcoder.py.1 # Remove any existing versions
!wget -O unixcoder.py https://raw.githubusercontent.com/microsoft/CodeBERT/master/UniXcoder/unixcoder.py

import torch
from unixcoder import UniXcoder

device = torch.device("cuda" if torch.cuda.is_available() else "cpu")
model = UniXcoder("microsoft/unixcoder-base")
model.to(device)

def get_embedding(code_str):
    tokens = model.tokenize([code_str])
    with torch.no_grad():
        emb = model(torch.tensor(tokens))[0]          # (1, seq_len, hidden)
    vec = emb.mean(dim=1).squeeze()     # mean pooling
    return vec

# =========================================================
# 1. PYTHON CODE A
# =========================================================
python_code_a = """
def add(a, b):
    return a + b
"""

# =========================================================
# 2. PYTHON CODE B
# =========================================================
python_code_b = """
def sum_two_numbers(x, y):
    return x + y
"""

# =========================================================
# 3. JAVA CODE A
# =========================================================
java_code_a = """
public int add(int a, int b) {
    return a + b;
}
"""

# =========================================================
# 4. JAVA CODE B
# =========================================================
java_code_b = """
public int sumTwoNumbers(int x, int y) {
    return x + y;
}
"""

# =========================================================
# 5. CREATE UNIxCODER EMBEDDINGS / VECTOR REPRESENTATIONS
# =========================================================
vec_py_a = get_embedding(python_code_a)
vec_py_b = get_embedding(python_code_b)

vec_java_a = get_embedding(java_code_a)
vec_java_b = get_embedding(java_code_b)

# =========================================================
# 6. CHECK VECTOR SHAPES / DIMENSIONS
# =========================================================
print("========== VECTOR SHAPES ==========")

print("Python A shape :", vec_py_a.shape)
print("Python B shape :", vec_py_b.shape)
print("Java A shape   :", vec_java_a.shape)
print("Java B shape   :", vec_java_b.shape)

print("\nPython A dimension :", vec_py_a.numel())
print("Python B dimension :", vec_py_b.numel())
print("Java A dimension   :", vec_py_b.numel())
print("Java B dimension   :", vec_java_b.numel())


# =========================================================
# 7. SHOW ACTUAL VECTOR VALUES
#    First 20 values are shown so output does not become huge
# =========================================================
print("\n========== ACTUAL VECTOR REPRESENTATIONS ==========")

print("\nPython A vector (first 20 values):")
print(vec_py_a[:20])

print("\nPython B vector (first 20 values):")
print(vec_py_b[:20])

print("\nJava A vector (first 20 values):")
print(vec_java_a[:20])

print("\nJava B vector (first 20 values):")
print(vec_java_b[:20])


# =========================================================
# 8. COSINE SIMILARITY: PYTHON A <-> PYTHON B
# =========================================================
sim_python = torch.nn.functional.cosine_similarity(
    vec_py_a.unsqueeze(0),
    vec_py_b.unsqueeze(0)
)

# =========================================================
# 9. COSINE SIMILARITY: JAVA A <-> JAVA B
# =========================================================
sim_java = torch.nn.functional.cosine_similarity(
    vec_java_a.unsqueeze(0),
    vec_java_b.unsqueeze(0)
)

# =========================================================
# 10. CROSS-LANGUAGE: PYTHON A <-> JAVA A
# =========================================================
sim_cross = torch.nn.functional.cosine_similarity(
    vec_py_a.unsqueeze(0),
    vec_java_a.unsqueeze(0)
)

# =========================================================
# 11. PRINT RESULTS
# =========================================================
print("\n========== COSINE SIMILARITY ==========")

print("Python A ↔ Python B similarity :", sim_python.item())
print("Java A ↔ Java B similarity     :", sim_java.item())
print("Python A ↔ Java A similarity    :", sim_cross.item())


# =========================================================
# 12. SAVE ACTUAL VECTOR REPRESENTATIONS
# =========================================================
torch.save({
    "python_a": vec_py_a,
    "python_b": vec_py_b,
    "java_a": vec_java_a,
    "java_b": vec_java_b
}, "/content/unixcoder_vectors.pt")

print("\nSaved vector file:")
print("/content/unixcoder_vectors.pt")
```
